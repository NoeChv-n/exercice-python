```python
import ssl
import seaborn as sns

ssl._create_default_https_context = ssl._create_unverified_context
df = sns.load_dataset("titanic")
print(df.head())
```

```python
nb_moins_de_18 = df[df["age"]<18].shape[0]
print(f"Moins de 18 ans : {nb_moins_de_18}")
```

```python
billets_les_plus_chers = df[["fare","class","embark_town"]].sort_values(by="fare",ascending=False).head(5)
print("Billets les plus chers :")
print(billets_les_plus_chers)
```

```python
femmes_1ere = df[(df["class"] == "First")&(df["sex"] == "female")].shape[0]/df[df["class"] == "First"].shape[0]
print(f"Femmes en 1ère classe : {femmes_1ere:.2%}")
```

```python
taux_survie_femmes = df[(df["survived"] == 1) & (df["sex"] == "female") & (df["class"]== "First")].shape[0] / df[(df["sex"] == "female") & (df["class"] == "First")].shape[0]
print(f"Survie des femmes en 1re : {taux_survie_femmes:.2%}")
taux_survie_hommes = df[(df["survived"] == 1) & (df["sex"] == "male") & (df["class"]== "Third")].shape[0] / df[(df["sex"] == "male") & (df["class"] == "Third")].shape[0]
print(f"Survie des hommes en 1re : {taux_survie_hommes:.2%}")
```
